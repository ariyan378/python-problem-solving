```python
import pandas as pd

# DataFrame 1: Customers
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104],
    "name": ["Alice", "Bob", "Charlie", "David"],
    "city": ["Dhaka", "Chittagong", "Khulna", "Sylhet"]
})

customers
```

```python
# DataFrame 3: Payments
payments = pd.DataFrame({
    "order_id": [1001, 1002, 1005],
    "payment_method": ["Card", "Bkash", "Cash"],
    "amount": [80000, 30000, 500]
})

payments
```

```python
# DataFrame 2: Orders
orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004],
    "customer_id": [101, 102, 101, 105],
    "product": ["Laptop", "Phone", "Keyboard", "Mouse"]
})

orders
```

```python
#1. Show all customer names with their ordered products
pd.merge(customers , orders ,on="customer_id",how = "inner" )[['name','product']]
```

```python
# 2. Find customers who did not place any order
no_order = pd.merge(customers , orders ,on="customer_id",how = "left" )[['name','product']]
no_order[no_order['product'].isnull()]
```

```python
# 3. Find orders that do not have customer information
third = pd.merge(customers , orders ,on="customer_id",how = "right" )[['customer_id','name','product']]
third[third['name'].isna()]
```

```python
#4. Show all orders with payment amount
pd.merge(orders, payments, on='order_id', how='inner')[
    ['order_id', 'product', 'amount']
]
```

```python
#5. Find products that are not paid yet
five = pd.merge(orders, payments, on='order_id', how='left')[
    ['order_id', 'product', 'amount']
]
five[five['amount'].isna()]
```
