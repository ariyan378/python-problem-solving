```python
df1_melt = df1.melt(id_vars="Name",
                  var_name="Year",
                  value_name="Courses")
df1_melt
```

```python

df3 = pd.DataFrame({"Name": ["Tom", "Tom", "Mike", "Mike"],
                   "Department": ["CS", "STATS", "CS", "STATS"],
                   "2018": [1, 2, 3, 1],
                   "2019": [2, 3, 4, 2],
                   "2020": [5, 1, 2, 2]}).melt(id_vars=["Name", "Department"], var_name="Year", value_name="Courses")
df3
```

# Merge DataFrame 

```python
df4 = pd.DataFrame({'A': [1, 3, 5],
                    'B': [2, 4, 6]})
df5 = pd.DataFrame({'A': [7, 9, 11],
                    'B': [8, 10, 12]})
```

```python
pd.concat((df4,df5),axis=0,ignore_index=True)
```

```python
df6 = pd.DataFrame({"name": ['Magneto', 'Storm', 'Mystique', 'Batman', 'Joker', 'Catwoman', 'Hellboy'],
                    'alignment': ['bad', 'good', 'bad', 'good', 'bad', 'bad', 'good'],
                    'gender': ['male', 'female', 'female', 'male', 'male', 'female', 'male'],
                    'publisher': ['Marvel', 'Marvel', 'Marvel', 'DC', 'DC', 'DC', 'Dark Horse Comics']})
df7= pd.DataFrame({'publisher': ['DC', 'Marvel', 'Image'],
                    'year_founded': [1934, 1939, 1992]})
```

```python
df6.merge(df7,on='publisher',how='outer',indicator=True)
```

```python
from datetime import datetime as dt

date = dt.strptime("July 9 2005, 13:54", "%B %d %Y, %H:%M")

date
```

```python
print(f"Year: {date.strftime('%Y')}")
print(f"Month: {date.strftime('%B')}")
print(f"Day: {date.strftime('%d')}")
print(f"Day name: {date.strftime('%A')}")
print(f"Day of year: {date.strftime('%j')}")
print(f"Time of day: {date.strftime('%p')}")
```
